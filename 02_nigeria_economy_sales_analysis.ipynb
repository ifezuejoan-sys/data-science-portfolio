```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": [
        "Jan 2026", "Feb 2026", "Mar 2026",
        "Apr 2026", "May 2026", "Jun 2026",
        "Jul 2026", "Aug 2026", "Sep 2026"
    ],
    "Headline_Inflation": [
        15.10, None, 15.38, None, None,
        15.91, 15.43, 15.39, 14.63
    ],
    "Food_Inflation": [
        8.89, None, 14.31, None, None,
        17.52, 20.31, 19.57, 19.59
    ]
}

df = pd.DataFrame(data)

display(df)

df.plot(
    x="Month",
    y=["Headline_Inflation", "Food_Inflation"],
    marker="o",
    figsize=(10, 5)
)

plt.title("Nigeria: Headline vs Food Inflation, 2026")
plt.ylabel("Year-on-year inflation (%)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calculate average inflation, ignoring missing values
print("Average headline inflation:",
      round(df["Headline_Inflation"].mean(), 2), "%")

print("Average food inflation:",
      round(df["Food_Inflation"].mean(), 2), "%")

# Compare the first and latest available headline figures
valid_headline = df.dropna(subset=["Headline_Inflation"])

change = (
    valid_headline["Headline_Inflation"].iloc[-1]
    - valid_headline["Headline_Inflation"].iloc[0]
)

print("Change in headline inflation:",
      round(change, 2), "percentage points")

# Find the month with the highest recorded food inflation
valid_food = df.dropna(subset=["Food_Inflation"])
highest = valid_food.loc[valid_food["Food_Inflation"].idxmax()]

print("Highest recorded food inflation:",
      highest["Food_Inflation"], "% in", highest["Month"])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep"],
    "Units_Sold": [500, 470, 420, 450, 390, 410, 380, 400, 370],
    "Unit_Price_NGN": [75000, 78000, 42000, 82000, 45000, 85000, 48000, 88000, 50000]
})

sales["Revenue_NGN"] = sales["Units_Sold"] * sales["Unit_Price_NGN"]

display(sales)

print("Total revenue: ₦{:,.0f}".format(sales["Revenue_NGN"].sum()))
print("Total units sold:", sales["Units_Sold"].sum())

plt.figure(figsize=(9, 5))
plt.plot(sales["Month"], sales["Revenue_NGN"], marker="o")
plt.title("Illustrative Monthly Sales Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (NGN)")
plt.grid(True)
plt.show()
```

```python
# Calculate month-to-month revenue growth
sales["Revenue_Growth_%"] = (
    sales["Revenue_NGN"].pct_change() * 100
).round(2)

# Calculate average selling price per unit
sales["Average_Selling_Price_NGN"] = (
    sales["Revenue_NGN"] / sales["Units_Sold"]
).round(2)

display(sales)

print("Highest revenue month:",
      sales.loc[sales["Revenue_NGN"].idxmax(), "Month"])

print("Lowest revenue month:",
      sales.loc[sales["Revenue_NGN"].idxmin(), "Month"])

print("Total revenue: ₦{:,.0f}".format(sales["Revenue_NGN"].sum()))
print("Average monthly revenue: ₦{:,.0f}".format(
    sales["Revenue_NGN"].mean()
))
```
